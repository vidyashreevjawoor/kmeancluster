## k-means cluster

##### identifies k cluster  (grouping)within the dataset

###### steps:

> 1)Choose 3(based on k value)  random points as  Centroids(mean)

> 2) Calculate distance bw points and centroids and than assign points to cloest centroid ..

> compute new centroids ..

>Note:  we need to repeat step 2 and step 3 until we reach stability (centroids value gets repeated)

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
iris.data
```

```python
iris.feature_names
```

```python
columns = iris.feature_names
```

```python
columns
```

```python
pd.DataFrame(iris.data,columns=iris.feature_names)
```

```python
df = pd.DataFrame(iris.data, columns=columns)
```

```python
from sklearn.model_selection import train_test_split
train ,test = train_test_split(df, test_size=0.2, random_state=42)
```

```python
train.shape
```

```python
test.shape
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=3,random_state=42)
```

```python
model.fit(train)
```

```python
predict = model.predict(test)
```

```python
pd.Series(predict).unique()
```
